## 6.4 Revenue by Location

### Business Question

Which location generates the highest revenue?

---

### Why are we doing this?

Knowing which location generates the most revenue helps the business:

- Identify the best-performing markets.
- Allocate marketing budgets effectively.
- Plan business expansion.
- Improve sales strategies in low-performing regions.

---

### Theory

Unlike the previous analysis, we are not counting customers.

Here, we calculate the total revenue generated by customers in each location.

To achieve this, we use Pandas `groupby()` and `sum()` functions.

- `groupby()` groups the dataset by a column.
- `sum()` calculates the total revenue for each group.

This is one of the most commonly used techniques in real-world data analysis.

```python
# Calculate total revenue for each location

location_revenue = df.groupby("Location")["TotalAmountSpent"].sum()

# Display the result
location_revenue
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    location_revenue.index,
    location_revenue.values,
    edgecolor="black"
)

# Display revenue above each bar
for i, value in enumerate(location_revenue.values):
    plt.text(i, value + 200, f"${value:,.0f}", ha="center")

plt.title("Revenue by Location", fontsize=14)

plt.xlabel("Location")

plt.ylabel("Total Revenue")

plt.savefig("images/revenue_by_location.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Output

A bar chart showing the total revenue generated by each location.

---

### Interpretation

The chart compares revenue across different locations.

Some cities may have fewer customers but generate higher revenue because customers spend more on average.

---

### Business Insight

Customer count alone does not determine business performance.

Revenue analysis helps identify the most profitable markets and highlights locations where customers have higher purchasing power.

---

### Business Recommendation

Continue investing in high-revenue locations while investigating why lower-performing locations generate less revenue. Marketing campaigns, product offerings, and pricing strategies can be adjusted to improve sales in those regions.


value_counts() is used to count how many times each category appears.
groupby() is used when we want to perform calculations (such as sum, mean, max, min, or count) on grouped data.

In this case, we wanted the total revenue for each location, so groupby() with sum() is the correct approach.

## 6.6 Revenue by Age Group

### Business Question

### Why are we doing this?

Customers of different age groups have different purchasing behaviors.

Understanding revenue by age group helps the business:

- Identify the most valuable customer segment.
- Design age-specific marketing campaigns.
- Recommend products based on customer age.
- Improve customer targeting.

---

### Theory

Age is a continuous numerical variable.

Instead of analyzing every individual age, we divide customers into age groups using the `pd.cut()` function.

Then we use:

- `groupby()` to group customers by age category.
- `sum()` to calculate total revenue for each age group.

```python
# Create age groups

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)

# Display first few rows
df[["Age", "AgeGroup"]].head()
```

```python
# Calculate revenue by age group

agegroup_revenue = df.groupby("AgeGroup")["TotalAmountSpent"].sum()

# Display the result
print(agegroup_revenue)
```

```python
# Calculate revenue by age group

agegroup_revenue = df.groupby("AgeGroup")["TotalAmountSpent"].sum()

# Display the result
plt.figure(figsize=(8,5))

plt.bar(
    agegroup_revenue.index.astype(str),
    agegroup_revenue.values,
    edgecolor="black"
)

# Display revenue above each bar
for i, value in enumerate(agegroup_revenue.values):
    plt.text(i, value + 200, f"${value:,.0f}", ha="center")

plt.title("Revenue by Age Group", fontsize=14)

plt.xlabel("Age Group")

plt.ylabel("Total Revenue")

plt.savefig("images/revenue_by_age_group.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Output

A bar chart showing the total revenue generated by each age group.

---

### Interpretation

The chart compares revenue across different customer age groups.

The age group with the tallest bar contributes the highest revenue.

---

### Business Insight

Different age groups have different purchasing behaviors.

The highest-revenue age group represents the company's most valuable customer segment and can guide future marketing and product strategies.

## 6.8 Customer Segmentation by Spending

### Business Question

How can customers be categorized based on their spending behavior?

---

### Why are we doing this?

Not all customers spend the same amount.

Customer segmentation helps businesses:

- Identify high-value customers.
- Create personalized marketing campaigns.
- Improve customer retention.
- Allocate marketing budgets effectively.
- Increase overall revenue.

---

### Theory

Customer segmentation divides customers into different groups based on their spending.

In this project, customers are categorized into three spending levels:

- Low Spenders
- Medium Spenders
- High Spenders

We use the `pd.cut()` function to create these categories based on the TotalAmountSpent column.

```python
#Step 1: Create Spending Categories

# Create customer spending segments

df["SpendingCategory"] = pd.cut(
    df["TotalAmountSpent"],
    bins=[0, 1000, 3000, 5000],
    labels=["Low", "Medium", "High"]
)

# Display first few rows
df[["CustomerID", "TotalAmountSpent", "SpendingCategory"]].head()
```

```python
#Step 2: Count Customers in Each Category
# Count customers in each spending category

df["SpendingCategory"] = pd.cut(
    df["TotalAmountSpent"],
    bins=[0, 1000, 3000, 5000],
    labels=["Low", "Medium", "High"]
)

# Display first few rows
df[["CustomerID", "TotalAmountSpent", "SpendingCategory"]].head()
segment_count = df["SpendingCategory"].value_counts()

print(segment_count)
```

```python
#Step 3: Create Bar Chart

plt.figure(figsize=(7,5))

plt.bar(
    segment_count.index.astype(str),
    segment_count.values,
    edgecolor="black"
)

# Display customer count above each bar
for i, value in enumerate(segment_count.values):
    plt.text(i, value + 0.3, str(value), ha="center")

plt.title("Customer Segmentation by Spending", fontsize=14)

plt.xlabel("Spending Category")

plt.ylabel("Number of Customers")

plt.savefig("images/customer_segmentation_bar.png", dpi=300, bbox_inches="tight")
plt.show()
```
